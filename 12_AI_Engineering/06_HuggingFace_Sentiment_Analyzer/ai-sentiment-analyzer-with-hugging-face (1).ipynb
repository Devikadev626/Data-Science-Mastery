```python
from transformers import pipeline
import gradio as gr
```

```python
#Load a Hugging Face Model

sentiment_pipeline = pipeline(
    "sentiment-analysis"
)
```

```python
#Test the Model
result = sentiment_pipeline(
    "I really enjoyed this course!"
)

print(result)
```

```python
#Test Multiple Inputs

texts = [
    "I really enjoyed this course!",
    "This product is terrible.",
    "The movie was okay.",
    "I absolutely love this application."
]

for text in texts:

    result = sentiment_pipeline(text)

    print("Text:", text)
    print("Result:", result)
    print()
```

```python
#Create an AI Function

def analyze_sentiment(text):

    if not text.strip():
        return "Please enter some text."

    result = sentiment_pipeline(text)[0]

    label = result["label"]
    score = result["score"]

    return (
        f"Sentiment: {label}\n"
        f"Confidence: {score:.2%}"
    )
```

```python
print(
    analyze_sentiment(
        "I love learning AI!"
    )
)
```

```python
import gradio as gr
```

```python
#Create the Interface

demo = gr.Interface(
    fn=analyze_sentiment,
    inputs=gr.Textbox(
        label="Enter your text",
        placeholder="Type something here..."
    ),
    outputs=gr.Textbox(
        label="Sentiment"
    ),
    title="Hugging Face Sentiment Analyzer",
    description="Enter text and let the AI predict its sentiment."
)
```

```python
#Launch the App

demo.launch()
```

```python
# Improve the UI


demo = gr.Interface(
    fn=analyze_sentiment,

    inputs=gr.Textbox(
        lines=5,
        label="Enter Text",
        placeholder="Write your sentence here..."
    ),

    outputs=gr.Textbox(
        label="AI Prediction"
    ),

    title="Hugging Face AI Sentiment Analyzer",

    description=(
        "Enter a sentence and the Hugging Face "
        "AI model will predict its sentiment."
    ),

    examples=[
        ["I love this product!"],
        ["This is an amazing experience."],
        ["I am disappointed with the service."],
        ["The product is okay."]
    ]
)
```

```python
demo.launch()
```
